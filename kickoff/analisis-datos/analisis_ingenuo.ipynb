# Rendimiento de cultivos por provincia — primer análisis

Objetivo: calcular el **rendimiento promedio de cultivos (soja y maíz) por
provincia**, a partir de un dataset cargado por otro equipo.

> Dataset ficticio con fines educativos — ver el README del proyecto.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/rendimiento_cultivos.csv")
df.head()
```

## Vistazo rápido al dataset

```python
print(df.shape)
df.info()
```

## Normalizar unidades

El dataset trae rendimientos en dos unidades distintas (`kg/ha` y `tn/ha`).
Para poder comparar todo en la misma escala, lo más directo es llevar todo
a `kg/ha`: si una fila dice `tn/ha`, multiplicamos por 1000.

```python
df["unidad"].value_counts()
```

```python
def a_kg_ha(row):
    if row["unidad"] == "tn/ha":
        return row["rendimiento"] * 1000
    return row["rendimiento"]

df["rendimiento_kg_ha"] = df.apply(a_kg_ha, axis=1)
df[["provincia", "campaña", "cultivo", "rendimiento", "unidad", "rendimiento_kg_ha"]].head()
```

## Rendimiento promedio por provincia

```python
promedio_por_provincia = (
    df.groupby("provincia")["rendimiento_kg_ha"]
    .mean()
    .sort_values(ascending=False)
)
promedio_por_provincia
```

```python
promedio_por_provincia.plot(kind="bar", figsize=(10, 5), title="Rendimiento promedio por provincia (kg/ha)")
plt.ylabel("kg/ha")
plt.tight_layout()
plt.show()
```
